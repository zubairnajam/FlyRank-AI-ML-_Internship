## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
#A single if-statement rule (e.g. "flag pages with impressions ≥ 500 AND days-since-update ≥ 180") only checks a couple of signals in isolation. Real decline shows up as a combination of weak signals — position drift, thinness, staleness, low CTR relative to position tier — none of which is decisive alone, but together they're predictive. That's exactly why the starter pipeline's own baseline rule tops out at Precision@50 = 0.240 while a random forest trained on the same signals reaches 0.740: the rule can't weigh and combine dozens of features the way a model can. It's not that the rule is wrong, it's that it's too rigid to capture interactions between signals — which is the textbook case for ML over a fixed rule.
```
